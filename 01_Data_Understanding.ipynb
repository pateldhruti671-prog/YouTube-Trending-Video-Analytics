```python
# Create a backup before cleaning
df_clean = df.copy()

print("Backup created successfully!")
print("Original rows:", len(df))
```

```python
# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

print("Rows after removing duplicates:", len(df_clean))
print("Duplicates remaining:", df_clean.duplicated().sum())
```

```python
missing_after_duplicates = df_clean.isnull().sum()

print("Missing values after removing duplicates:")
print(missing_after_duplicates)
```

```python
# Fill missing descriptions with an empty string
df_clean['description'] = df_clean['description'].fillna('')

print("Missing descriptions after cleaning:",
      df_clean['description'].isnull().sum())

print("Total rows:", len(df_clean))
```

```python
# Convert date columns to datetime format

df_clean['trending_date'] = pd.to_datetime(
    df_clean['trending_date'],
    format='%y.%d.%m'
)

df_clean['publish_time'] = pd.to_datetime(
    df_clean['publish_time'],
    utc=True
)

print("Date conversion completed!")
print(df_clean[['trending_date', 'publish_time']].dtypes)
```

```python
# Calculate engagement rate

df_clean['engagement_rate'] = (
    (df_clean['likes'] + df_clean['comments_count'] if 'comments_count' in df_clean.columns else df_clean['likes'] + df_clean['comment_count'])
    / df_clean['views']
) * 100
```

```python
# Calculate engagement rate
df_clean['engagement_rate'] = (
    (df_clean['likes'] + df_clean['comment_count'])
    / df_clean['views']
) * 100

print("Engagement rate created successfully!")
```

```python
# Extract publishing hour
df_clean['publish_hour'] = df_clean['publish_time'].dt.hour

print("Publish hour created successfully!")
```

```python
# Extract day of the week
df_clean['publish_day'] = df_clean['publish_time'].dt.day_name()

print("Publish day created successfully!")
```

```python
print("Dataset shape:", df_clean.shape)

print("\nNew columns:")
print(df_clean[['engagement_rate', 'publish_hour', 'publish_day']].head())

print("\nMissing values:")
print(df_clean.isnull().sum().sum())
```

```python
trending_duration = (
    df_clean.groupby('video_id')['trending_date']
    .agg(['min', 'max', 'count'])
    .reset_index()
)

trending_duration.columns = [
    'video_id',
    'first_trending_date',
    'last_trending_date',
    'trending_records'
]

print(trending_duration.head())
```

```python
trending_duration['trending_duration_days'] = (
    trending_duration['last_trending_date']
    - trending_duration['first_trending_date']
).dt.days + 1

print(trending_duration.head())
```

```python
longest_trending = (
    trending_duration
    .sort_values('trending_duration_days', ascending=False)
    .head(10)
)

print(longest_trending)
```

```python
print(trending_duration['trending_duration_days'].describe())
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    trending_duration['trending_duration_days'],
    bins=30
)

plt.title('Distribution of YouTube Trending Duration - India')
plt.xlabel('Trending Duration (Days)')
plt.ylabel('Number of Videos')

plt.tight_layout()
plt.show()
```

```python
print(trending_duration['trending_duration_days'].describe())
```

```python
video_details = (
    df_clean[
        ['video_id', 'title', 'channel_title', 'category_name',
         'views', 'likes', 'comment_count']
    ]
    .drop_duplicates('video_id')
)

trending_analysis = trending_duration.merge(
    video_details,
    on='video_id',
    how='left'
)

print("Rows:", len(trending_analysis))
print(trending_analysis.head())
```

```python
long_trending = (
    trending_analysis[
        trending_analysis['trending_duration_days'] >= 7
    ]
    .sort_values('trending_duration_days', ascending=False)
)

print("Videos trending for 7+ days:",
      len(long_trending))

print(long_trending[
    ['title', 'channel_title', 'category_name',
     'trending_duration_days', 'views']
].head(20))
```

```python
long_trending_categories = (
    long_trending['category_name']
    .value_counts()
)

print(long_trending_categories)
```

```python
plt.figure(figsize=(10, 6))

long_trending_categories.head(10).sort_values().plot(
    kind='barh'
)

plt.title('Categories with Most Videos Trending for 7+ Days')
plt.xlabel('Number of Videos')
plt.ylabel('Category')

plt.tight_layout()
plt.show()
```

```python
duration_groups = pd.cut(
    trending_analysis['trending_duration_days'],
    bins=[0, 1, 3, 7, 30, float('inf')],
    labels=[
        '1 Day',
        '2–3 Days',
        '4–7 Days',
        '8–30 Days',
        '30+ Days'
    ]
)

duration_summary = duration_groups.value_counts().sort_index()

print(duration_summary)
```

```python
plt.figure(figsize=(10, 6))

duration_summary.plot(kind='bar')

plt.title('Distribution of YouTube Trending Duration')
plt.xlabel('Trending Duration')
plt.ylabel('Number of Videos')

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
